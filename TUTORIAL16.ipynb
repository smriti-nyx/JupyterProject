FILTER FUNCTION IN PYTHON

```python
def even(n):
    if n%2 == 0 :
        return True
```

```python
lst = [ 1,2,3,4,5,6,7,8,9,0]
```

```python
list(filter(even, lst)) #filter( fnc name , iterable)
```

```python
#converting it into a lambda function.
```

```python
list(filter(lambda n : n%2 == 0 , lst))
```

```python
#converting into map fnc.
list(map(lambda n : n%2 == 0 , lst))
```
