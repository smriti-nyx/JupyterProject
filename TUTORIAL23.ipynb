CUSTOM EXCEPTION HANDLING IN PYTHON

```python
class Error (Exception):
    pass
class DobException(Error):
    pass
```

```python
year = int(input("Enter the year of your birth: "))
age = 2026 - year
try :
    if age <= 30 & age > 20 :
        print("Hola , you are eligible .")
        pass
    else :
        raise DobException
except DobException :
    print("Your age does not lie in the right range. ")
```
