```python
#Closures
def main_welcome():
    msg = "Hello everyone"
    def sub_welcome_class():
        print("Welcome to sera's world")
        print(msg)
        print("Have a good time. ")
    return sub_welcome_class()
```

```python
main_welcome()
```

```python
#Closures
def main_welcome(func):
    msg = "Hello everyone"
    def sub_welcome_class():
        print("Welcome to sera's world")
        func(msg)
        print("Have a good time. ")
    return sub_welcome_class()
```

```python
main_welcome(print)
```

```python
#Closures
def main_welcome(func):

    def sub_welcome_class():
        print("Welcome to sera's world")
        print(func([1,2,3,4,5,6,7]))
        print("Have a good time. ")
    return sub_welcome_class()
```

```python
main_welcome(len)
```

```python
#Closures
def main_welcome(func):

    def sub_welcome_class():
        print("Welcome to sera's world")
        func()
        print("Have a good time. ")
    return sub_welcome_class()
```

```python
def channel_name():
    print("This is Sera ")
```

```python
main_welcome(channel_name)
```

```python
#to directly use channel name as a parameter for the main wwelcome function is to add @main_welcome
```

```python
@main_welcome
def channel_name():
    print("This is Sera ")
```
