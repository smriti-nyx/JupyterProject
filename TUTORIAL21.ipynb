OOPS IN PYTHON

```python
class Car: #initialisation  #object
    def __init__ ( self, window , door , enginetype):
        self.window = window
        self.door = door
        self.enginetype = enginetype
        #function inside here is called method, says bout the method ofo the object.
    def self_driving(self ):
        return "This is a {} car. ".format(self.enginetype)
```

```python
car1 = Car(4,5,"petrol")
car2 = Car(3, 4 ,"diesel")
```

```python
print(car1.window)
print(car2.enginetype)

```

```python
print(car1.self_driving())
```
