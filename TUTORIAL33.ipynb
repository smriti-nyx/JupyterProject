 OOPS- MULTIPLE INHERITANCE IN PYTHON

```python
class A:
    def method1(self):
        print("A class method is called. ")


```

```python
class B(A):
    def method1(self):
        print("B class method is called. ")
        super().method1()
    def method2(self):
        print("B class method2 is called. ")


```

```python
class C(A):
    def method1(self):
        print("C class method is called. ")
        super().method1()
```

```python
class D(B,C):
    def method1(self):
        print("D class method is called. ")
        super().method1()
```

```python
d = D()
```

```python
d.method1()
```
