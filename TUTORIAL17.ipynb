List Comprehension

```python
#list comprehension
```

```python
#the usual way
def square_lst(lst):
    lst1 = []
    for i in lst:
        lst1.append(i * i)
    return lst1
```

```python
square_lst([1,2,3,4,5,6,7,8])
```

```python
lst = [ 1,2,3,4,5,6,7,8]
```

```python
#the comprehended way

[i*i for i in lst]
```

```python
##
[i*i for i in lst if i%2 == 0]
```

```python
#storing the result in a variable
lst1 = [i*i for i  in lst if i%2 != 0 ]
```
