FUNCTIONS IN PYTHON

```python
#defining a usual function
def even_odd(num):
    if num%2 ==0:
        print("The number is even ")
    else:
        print("The number is odd")
```

```python
even_odd(90)
```

```python
#showing a positional argument
even_odd() #giving no parameter , the error will point the presence of a positional argument .
```

```python
def hello(name , age ):
    print("My name is {} . My age is {}. ".format(name , age))
```

```python
hello("Sera" , 25)
```

```python
hello()
#the error displays the term positional arguments , hence name and age were positional arguments .
```

```python
#positional argument - the parameter doesnt have key value pairs
##keyword argument - the parameter has key value pairs , like age = 29
def hello(name , age = 25):
    print("My name is {}. My age is {}.".format(name , age ))
```

```python
hello('Sera') #age is still getting printed even without putting it as a parameter here.
```

```python
hello()
#only one positional argument is missing as shown in the error msg .
```

```python
def hello(*args , **kwargs):
    print(args)
    print(kwargs)
```

```python
hello('Sera' , 'Woods' , age = 25 , year = 1800)
```

```python
lst = ['Sera', 'Woods'] #storing the positional args in a list
dict_args = { 'age' : 25 ,
              'year' : 1800} #and keyword argument in a dictionary
```

```python
hello(lst , dict_args) #without mentioning , everything has been printed as positional argument.
```

```python
hello(*lst , **dict_args) #fixed
```

```python
lst1 = [1, 2, 3, 4, 5, 6, 7]

def even_oddsum(lst1):
    even_sum = 0
    odd_sum = 0

    for i in lst1:
        if i % 2 == 0:
            even_sum = even_sum + i
        else:
            odd_sum = odd_sum + i

    return even_sum, odd_sum
```

```python
lst1 = [ 1,2,3,4,5,6,7]

even_oddsum(lst1)
```
