LAMBDA FUNCTION

```python
#it is a function with no name
#it works faster than a normal function
```

```python
#normal fnc.
def add(a, b ):
    return a+b
```

```python
#lambda fnc
# Function name = lambda parameter1, parameter2 : operation
#function(pass the inputs )
add= lambda a,b : a+b
```

```python
add(12,15)
```
