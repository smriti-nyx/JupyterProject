```python
#list as iterable , all elements of the list are allocated memory all at once.
lst = [ 1,2,3,4,5]
for i in lst :
    print(i)
```

```python
#list as iterator , #no values will be initialised initially . They will be initialised and allocated a memory once we call them .
lst1 = iter(lst)
```

```python
#using next fnc to call them.
lst1
```

```python
next(lst1)
```
