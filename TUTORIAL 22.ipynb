```python
#try and except
try :
    a = b
except Exception as ex : #general statemnet to print the kind of error
    print(ex)
```

```python
try :
    a = 10
    b = 0
    c = a/b
except ZeroDivisionError:
    print("Division by zero is not possible .")
except Exception as ex : #general statement to print the kind of error
    print(ex)
```
