```python
import seaborn as sns
import pandas as pd
df = sns.load_dataset("tips")
df.head()
```

```python
df.dtypes #dtypes is not a fnc but a property that already returns a pandas series therefore we need mot put a bracket() after it , it is used for parameters.
```

```python
#correlation with heatmap
```

```python
#uses colored cells, typically in a monochromatic scale to show a 2D matrix between two discrete dimensions or event types
```

```python
df.select_dtypes(include ="number").corr().head()
```

```python
sns.heatmap(df.select_dtypes(include="number").corr().head())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (15,8))
sns.heatmap(df.select_dtypes(include ="number").corr().head() , cmap = 'coolwarm', linecolor = 'y')
```

```python
#Jointplot
#helps to study the relation between 2 numeric values , usually a scatterplot, a hexbin plot , a 2D histogram or a 2D density plot
```

```python
#Univariate Analysis
```

```python
sns.jointplot(x = 'tip', y = 'total_bill', data = df,kind = 'hex')
```

```python
sns.jointplot(x = 'tip', y = 'total_bill', data = df,kind = 'reg')
```

```python
#Pairplot
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df.select_dtypes(include ="number").corr().head())
#pairplot for more than two independent variables
```

```python
sns.pairplot(df, hue = 'sex')
```

```python
sns.pairplot(df, hue = 'size')
```
