INHERITANCE OOPS

```python
class  Car :
    def __init__ (self, doors, windows, enginetype):
        self._doors = doors #protected
        self.__windows = windows #private
        self.enginetype = enginetype #public
        def drive(self):
            print("The person drives the car. ")
```

```python
class Audi (Car):
    def __init__ (self, windows,doors, enginetype ,enableai ):#new feature
        super().__init__(windows , doors , enginetype) #specifying that we got all these parameters from the parent class
        self.enableai = enableai #new parameter
        def selfdriving(self):
            print("Audi supports self driving .")
```

```python
audi1 = Audi( 5, 5 , "diesel",True)
```
