ITERATOR VS GENERATOR

```python
#iterable - the list lsst is iterable coz we can iterate through it
lst = [1,2,3,4,5,6,7]
for i in lst :
    print(i)
```

```python
#iterator- iter(next prints it ) creates and iterator , iterator produces values one by one
iterator = iter(lst)
```

```python
type(iterator)
```

```python
#we can iterate
for i in iterator :
    print(i)
```

```python
next(iterator)
```

```python
#uisng try n except(catch) method
try:
   print( next(iterator))
except StopIteration :
    print("The iterator is empty. ")
```

```python
#Generator  - easy way to create iterator
def square(n):
    for i in range(n):
        return i**2

```

```python
square(5) #iteration stopped at 1st elemnet that is 0
```

```python
#Generator  - easy way to create iterator
def square(n):
    for i in range(n):
        yield i**2

```

```python
for i in square(9):
    print(i)

```

```python
a = square(3)
a
```

```python
next(a)
```

difference
1. to create iterator , we use iter() and for generator ,w e use yield keyword
