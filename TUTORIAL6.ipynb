```python
from io import StringIO , BytesIO
import pandas as pd
import numpy as np
```

```python
StringIO() #File-like Interface: A StringIO object has methods that mimic a real file object, such as read(), write(), readline(), seek() (move the cursor position), and tell() (get the current cursor position).
#In-Memory Storage: The data is stored in a temporary buffer in your computer's memory (RAM), not on the hard drive. This makes operations generally faster than disk I/O, especially for small, temporary data manipulations.
```

```python
data1= ('col1,col2,col3 \n'
         '1,2,3 \n'
        '4,5,6 \n'
        '7,8,9\n')
```

```python
print(data1)
```

```python
df = pd.read_csv(StringIO(data1) , dtype = float)
print(df)
```

```python
df.dtypes
```

```python
df= pd.read_csv(StringIO(data1) , dtype = int)
print(df)
```

```python
df= pd.read_csv(StringIO(data1) , dtype = object) #object = strings
print(df)
```

```python
#using coordinates to get specific elements...i.e INDEXING
```

```python
df['col1']
```

```python
df['col2'][1]
```

```python
#applying diff datatypes to diff columns
```

```python
df= pd.read_csv(StringIO(data1) , dtype = {'col1' : int , 'col2' : float ,'col3' : object })
print(df)
#keep the dictionary format for applying diff data types to diff columns
```

```python
type(df['col2'][2])
```

```python
#checking the datatypes
df.dtypes

```

```python
#finding dtype of a specific element
type(df['col2'][2])


```

```python
#changing column values
df = pd.read_csv(StringIO(data1) , index_col =2) #decides which columns becomes the index that "0th column"
print(df)
```
