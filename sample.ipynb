```python
myset = {"bat", "aqua", "super","spider"}
print(myset)
type(myset)
```

```python
#Inbuilt functions in set
#addition of sets
myset = {"bat", "aqua", "super","spider"}
myset.add("hulk")
print(myset)
```

```python

#Difference of sets
myset = {"bat", "aqua", "super","spider"}
ourset = {"bat", "aqua", "super","spider", "hulk"}
ourset.difference(myset)
```

```python
ourset #ouset is not updated yet. to fix it , we use update fucntion
```

```python
myset = {"bat", "aqua", "super","spider"}
ourset = {"bat", "aqua", "super","spider", "hulk"}
ourset.difference_update(myset)
```

```python
myset = {"bat", "aqua", "super","spider"}
ourset = {"bat", "aqua", "super","spider", "hulk"}
ourset.difference_update(myset)
print(ourset)
```

```python
ourset#the ourset is updated now after using difference update fucntion.
```

```python
myset = {"bat", "aqua", "super","spider"}
ourset = {"bat", "aqua", "super","spider", "hulk"}
ourset.intersection(myset)
```

```python
#Printing the Values
mydict = { "carA" : "Audi" , "carB" : "Benz" , "carC" : "BMW" , "carC" : "Jeep" }
for x in mydict.values():
    print(x)
```
