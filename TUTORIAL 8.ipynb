```python
#defining and y values
x = np.arange( 0, 4* np.pi , 0.1)
y = np.sin(x)
plt.title('sine waveform')

#plotting the curve

plt.plot(x,y)
plt.show()
```

```python
#subplot  axis defined
x = np.arange(0, 5*np.pi , 0.2)
y_sin = np.sin(x)
y_cos = np.cos(x)

#plotting the first subplot
plt.subplot(2,2,1)
plt.title('sine curve')
plt.plot(x,y_sin)

#plotting the 2nd subplot
plt.subplot(2,2,2)
plt.title('cosine curve')
plt.plot(x, y_cos ,'y')

plt.show()
```

```python
#Bar plot

x = [2,4,6]
y = [11,6,7]

x2 = [ 3,9,11]
y2 = [6,15,7]

plt.xlabel('x axis ')
plt.ylabel('y axis')

plt.title('Bar Graph')
plt.bar(x,y , align = 'center')
plt.bar(x2 ,y2 ,color = 'g')
plt.show()
```

```python
#PIE CHART
```

```python
,#data to plot
labels = 'sera' ,'cera','serra', 'sarah'
colors =  ['gold' , 'yellowgreen' ,'lightcoral' ,'lightskyblue']
sizes = [ 10,20,39,5]
explode =[0.1,0,0,0.2 ]#exploding the slices ]

#plot
plt.pie(sizes , explode = explode , colors = colors, labels = labels,
autopct = '%1.1f%%' , shadow = True)

plt.axis('equal')
plt.show()
```
