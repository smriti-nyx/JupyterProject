Shallow  Copyy

```python
lst1 = [1,2,3,4]
lst2 = lst1.copy()
lst1, lst2
```

```python
lst2[1] = 100
lst1, lst2
```

```python
lst1 = [1, 2, 3, 4]
lst2 = lst1.copy()
lst1, lst2
```

```python
lst1[1] = 100
lst1,lst2
```

Shallow copy nested list

```python
lst1 = [ [ 1,2,3,4], [5,6,7,8]]
lst2 = lst1.copy()
```

```python
lst1[1][0] = 100
lst1,lst2
```

```python
lst1.append([2,3,5,8]) #no change in the og copy in the case of nested lists
```

deep copy

```python
##deep copy
import copy
lst1 = [1,2,3,4]
lst2 = copy.deepcopy(lst1)
lst1 , lst2
```

```python
lst2[1] = 100
lst1, lst2
```

in normal list , shallow copy = deep copy

```python
#nested lists (in deep copy)
import copy
lst1 = [[ 1,2,3,4] ,[6,7,8,9]]
lst2 = copy.deepcopy(lst1)
lst1, lst2
```

```python
lst2 = copy.deepcopy(lst1)
lst1, lst2

```

```python
lst2[1][1] = 100
lst1, lst2
```
