```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("Tips")
df.head()
```

```python
#Categorical plots
```

```python
# 1Countplot
```

```python
sns.countplot(x = 'sex', data = df)
```

```python
#2.barplot
```

```python
sns.barplot(x='sex', y = 'tip', data = df)
```

```python
sns.countplot(y ='day', data = df)
```

```python
# 3. barplot
```

```python
sns.boxplot(x = 'smoker' , y ='total_bill', data = df)
```

```python
sns.boxplot(x = 'smoker' , y ='total_bill', data = df , palette = 'rainbow')
```

```python
sns.boxplot(orient = 'v', data = df)
```

```python
sns.boxplot(x= "smoker", y= "total_bill" , hue = "day", data = df)
```

```python
sns.boxplot(x= "day", y= "total_bill" , hue = "smoker", data = df)
```

```python
#Violin plot
```

```python
sns.violinplot(  y ="total_bill", x = "day", data = df , palette = 'rainbow')
```
